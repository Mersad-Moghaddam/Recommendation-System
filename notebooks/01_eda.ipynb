# MovieLens exploratory analysis
This short notebook loads the same validated data used by the application and summarizes its scale, distribution, genres, and sparsity.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))
import matplotlib.pyplot as plt
from config import settings
from recommender.data import load_movielens
movies, ratings = load_movielens(settings.movies_csv, settings.ratings_csv)
```

```python
summary = {
    'movies': len(movies), 'users': ratings.userId.nunique(),
    'ratings': len(ratings), 'mean_rating': ratings.rating.mean(),
    'median_rating': ratings.rating.median(),
    'sparsity': 1 - len(ratings)/(ratings.userId.nunique()*movies.movieId.nunique()),
}
summary
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
ratings.rating.value_counts().sort_index().plot.bar(ax=axes[0], title='Rating distribution')
movies.genres.str.get_dummies('|').sum().sort_values().plot.barh(ax=axes[1], title='Genre frequency')
plt.tight_layout()
```
